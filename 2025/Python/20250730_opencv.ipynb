```python
import numpy as np

image = np.ones((512,512,3), dtype=np.uint8)
cv2.arrowedLine(image, (100,100), (400,400), (0,0,255), 3, line_type=cv2.LINE_AA, tipLength=0.2)
cv2.arrowedLine(image, (100,400), (400,100), (255,0,0), 3, line_type=cv2.LINE_AA, tipLength=0.1)
cv2.imshow('jinbo bosu arrowed', image)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
image = np.ones((512,512,3), dtype=np.uint8)*255
# 1
cv2.line(image, (10,10), (500,500), (255,0,0), 2)
# 2
cv2.line(image, (100,400), (400,400), (0,255,0), 2)
# 3
cv2.arrowedLine(image, (50,50), (300,50), (0,0,255), 2, line_type=cv2.LINE_AA, tipLength=0.2)
# 4
cv2.arrowedLine(image, (250,100), (250,400), (255,0,255), 2, line_type=cv2.LINE_AA, tipLength=0.1)
# 5
cv2.rectangle(image, (50,50), (300,200), (0,255,0), 2)
# 6
cv2.rectangle(image, (100,300,200,100), (255,0,0), -1)
# 7
cv2.circle(image, (256,256), 100, (255,0,0), -1)
# 8
cv2.circle(image, (400,100), 40, (0,255,255), 5)
cv2.imshow('jinbo arrowed', image)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
image = np.ones((512,512,3), dtype=np.uint8)
cv2.rectangle(image, (50,50), (250,250), (255,255,255), 2)
cv2.rectangle(image, (300,300,50,100), (255,255,255), 2)
cv2.imshow('jinbo rectangle', image)
cv2.waitKey()
cv2.destroyAllWindows()
```
