## 필터링

```python
import cv2
import numpy as np

src = cv2.imread('pier39.jpg', cv2.IMREAD_GRAYSCALE)
kernel = np.ones((3,3), dtype=np.float64)/9
dst = cv2.filter2D(src, -1, kernel)

cv2.imshow('src', src)
cv2.imshow('dst', dst)
cv2.waitKey()
cv2.destroyAllWindows()
```

## 블러링

```python
import matplotlib.pyplot as plt

img = cv2.imread('pier39.JPG')
plt.figure(figsize=(20,5))

for i in range(1,4):
    k = i*2+1
    dst = cv2.blur(img, (k,k))
    desc = f'k = {k}'
    cv2.putText(dst, desc, (10,30), cv2.FONT_HERSHEY_SIMPLEX, 1.0, 0, 1, cv2.LINE_AA)

    index = 140+i
    plt.subplot(index)
    plt.imshow(cv2.cvtColor(dst, cv2.COLOR_BGR2RGB))
    plt.axis('off')
plt.show()
```

## 가우시안 필터링

```python
img = cv2.imread('pier39.JPG')
plt.figure(figsize=(20,5))

for i in range(1,4):
    k = i*2+1
    dst = cv2.GaussianBlur(img, (k,k), i-1)
    desc = f'k = {k}'
    cv2.putText(dst, desc, (10,30), cv2.FONT_HERSHEY_SIMPLEX, 1., 0, 1, cv2.LINE_AA)

    index = 140+i
    plt.subplot(index)
    plt.imshow(cv2.cvtColor(dst, cv2.COLOR_BGR2RGB))
    plt.axis('off')
plt.show()
```

## 샤프닝

```python
img = cv2.imread('pier39.JPG', cv2.IMREAD_GRAYSCALE)
plt.figure(figsize=(15,5))

blur = cv2.GaussianBlur(img, (5,5), 0)
sharp = cv2.addWeighted(img, 1.5, blur, -0.5, 0)

plt.subplot(131),plt.imshow(img, 'gray'),plt.title('origin'),plt.axis('off')
plt.subplot(132),plt.imshow(blur, 'gray'),plt.title('blur'),plt.axis('off')
plt.subplot(133),plt.imshow(sharp, 'gray'),plt.title('unsharp'),plt.axis('off')
plt.show()
```

## 노이즈

### 양방향 필터

```python
img = cv2.imread('beach.jpg', cv2.IMREAD_GRAYSCALE)

dst = cv2.bilateralFilter(img, -1, 10, 5)
cv2.imshow('img', img)
cv2.imshow('dst', dst)

cv2.waitKey()
cv2.destroyAllWindows()
```

### 미디언 필터

```python
img = cv2.imread('lena-300x300.png')

dst = cv2.medianBlur(img, 5)
cv2.imshow('img', img)
cv2.imshow('dst', dst)

cv2.waitKey()
cv2.destroyAllWindows()
```

### 기하학적 변환

```python
src = cv2.imread('scenery.jpg')
aff = np.array([[1,0,100],[0,1,100]], dtype=np.float32)
dst = cv2.warpAffine(src, aff, (0,0))

cv2.imshow('src', src)
cv2.imshow('dst', dst)
cv2.waitKey()
cv2.destroyAllWindows()
```

### 영상 크기 변환

```python
src = cv2.imread('candies.png')
dst1 = cv2.resize(src, (0,0), fx=4, fy=4, interpolation=cv2.INTER_NEAREST)
dst2 = cv2.resize(src, (1920,1280))
dst3 = cv2.resize(src, (1920,1280), interpolation=cv2.INTER_CUBIC)
dst4 = cv2.resize(src, (1920,1280), interpolation=cv2.INTER_LANCZOS4)

cv2.imshow('src', src)
cv2.imshow('dst1', dst1[500:900, 400:800])
cv2.imshow('dst2', dst2[500:900, 400:800])
cv2.imshow('dst3', dst3[500:900, 400:800])
cv2.imshow('dst4', dst4[500:900, 400:800])

cv2.waitKey()
cv2.destroyAllWindows()
```

### 영상의 회전

```python
src = cv2.imread('scenery.jpg')
rad = 20*np.pi/180
aff = np.array([[np.cos(rad), np.sin(rad), 0],[-np.sin(rad), np.cos(rad), 0]], dtype=np.float32)
dst = cv2.warpAffine(src, aff, (0,0))

cv2.imshow('src', src)
cv2.imshow('dst', dst)
cv2.waitKey()
cv2.destroyAllWindows()
```
