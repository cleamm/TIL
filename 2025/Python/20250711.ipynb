```python
# sigmoid 함수
def sigmoid(x):
    return 1/(1 + np.exp(-x))
```

# 손실함수

```python
# 뉴런의 입력과 출력 정의
# 경사 하강법을 이용한 뉴런 학습이 손실을 줄이는 단계
x, y = 1, 0
w = np.random.normal(0,1, 1) # 가중치
output = sigmoid(x*w) # 출력값
print(x, y, '---->', output)
```

```python
# 에러 = 기대출력 - 출력
error = y - output
error
```

```python
# 경사 하강법을 이용한 뉴런 학습이 손실을 줄이는 단계
x, y = 1, 0 # 튜플 -> 패킹 -> 언패킹
learning_rate = 0.1

w = np.random.normal(0,1, 1)
print(w) # 랜덤 가중치
# 경사하강법
for i in range(1000):
    output = sigmoid(x * w)
    error = y - output
    # 오차에 따른 가중치 재계산
    w = w + x * learning_rate * error
    if i % 100 == 99:
        print(f'반복수:{i},\t 에러:{error},\t 출력:{output},\t 가중치:{w}')
```

```python
# x가 0일 때
x, y = 0, 1
learning_rate = 0.1

w = np.random.normal(0,1, 1)
for i in range(1000):
    output = sigmoid(x * w)
    error = y - output
    w = w + x * learning_rate * error
    if i % 100 == 99:
        print(f'반복수:{i},\t 에러:{error},\t 출력:{output},\t 가중치:{w}')
# 입력이 0이 들어가면 계산이 잘못됨
```

```python
# 편향을 추가하면...
x, y = 0, 1
learning_rate = 0.1
w = np.random.normal(0,1, 1)

# 편향을 추가
b = np.random.normal(0,1, 1)

for i in range(1000):
    output = sigmoid(x * w + 1 * b) # 편향을 첨가
    error = y - output
    w = w + x * learning_rate * error
    # 오차에 따른 편향 재계산
    b = b + 1 * learning_rate * error
    if i % 100 == 99:
        print(f'반복수:{i},\t 에러:{error},\t 출력:{output},\t 가중치:{w},\t 편향:{b}')
# 가중치는 변치 않으며 이외의 데이터는 변함
```

```python
# 첫번째 신경망 네트워크 : AND
x = np.array([[1,1], [1,0], [0,1], [0,0]])
y = np.array([[1], [0], [0], [0]])

w = np.random.normal(0,1,size=(2,))
b = np.random.normal(0,1,size=(1,))
```

```python
# 행우선
for i in x:
    for j in i:
        print(j)
```

AND 의 데이터를 준비하고 가중치와 편향을 초기화 한다.

```python
# w = np.random.normal(0,1,size=(1,)) 해당 코드로 하면 강의 내용과 결과가 다를 것임
# b = np.random.normal(0,1,size=(1,))
learning_rate = 0.1
for i in range(1000):
    error_sum = 0
    for j in range(4):
        output = sigmoid(np.sum(x[j] * w) + 1 * b)
        error = y[j][0] - output
        w = w + x[j] * learning_rate * error
        b = b + 1 * learning_rate * error
        error_sum += error
    if i % 100 == 99:
        print(f'반복수:{i},\t 에러:{error},\t 출력:{output},\t 가중치:{w}, 편향:{b}')
```

이렇게 학습한 네트워크가 정상적으로 동작하는지 확인해 보자.

```python
# AND 네트워크의 평가
for i in range(4):
    # w, b는 모델이 학습한 마지막 가중치와 편향
    sum_ = np.sum(x[i]*w) + b # 출력(예측한 정답)
    print('X:',x[i], '---> y:', y[i], 'predict:',sigmoid(sum_))
```

실제 논리대수 OR 연산에 신경망을 적용해 보자

```python
# 두번째 신경망 네트워크 : OR
x = np.array([[1,1], [1,0], [0,1], [0,0]])
y = np.array([[1], [1], [1], [0]])

w = np.random.normal(0,1,size=(2,))
b = np.random.normal(0,1,size=(1,))

learning_rate = 0.1
for i in range(1000):
    error_sum = 0
    for j in range(4): # 행
        output = sigmoid(np.sum(x[j] * w) + 1 * b)
        error = y[j][0] - output
        w = w + x[j] * learning_rate * error
        b = b + 1 * learning_rate * error
        error_sum += error
    if i % 100 == 99:
        print(f'반복수:{i},\t 에러:{error},\t 출력:{output},\t 가중치:{w}, 편향:{b}')
```

```python
# OR 네트워크의 평가
for i in range(4):
    # w, b는 모델이 학습한 마지막 가중치와 편향
    sum_ = np.sum(x[i]*w) + b # 출력(예측한 정답)
    print('X:',x[i], '---> y:', y[i], 'predict:',sigmoid(sum_))
```
