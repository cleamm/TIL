# 2. 분류문제

    수업중 진행한 “07지도학습KNN”, “07지도학습SVM” 에서 사용한 scikit 의 load_wine, load_breast_cancer 데이터를 DecisionTree 로 분류하고 평가지표 정확도, classification_report 출력


```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_wine, load_breast_cancer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split
import pandas as pd
import numpy as np
wine, cancer = load_wine(), load_breast_cancer()
df_wine = pd.DataFrame(wine.data, columns=wine.feature_names)
df_cancer = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df_wine.head()
```

```python
# wine.target_names, wine.target
# cancer.target_names, cancer.target
```

```python
# df_wine.head()
# df_cancer.head()
```

```python
c_X_train, c_X_test, c_y_train, c_y_test = train_test_split(df_cancer, cancer.target, test_size=.3, random_state=42)
w_X_train, w_X_test, w_y_train, w_y_test = train_test_split(df_wine, wine.target, test_size=.3, random_state=42)
print(c_X_train.shape, c_y_train.shape)
print(w_X_train.shape, w_y_train.shape)
```

```python
dt_cancer = DecisionTreeClassifier()
dt_wine = DecisionTreeClassifier()
```

```python
%%time
dt_cancer.fit(c_X_train, c_y_train)
dt_wine.fit(w_X_train, w_y_train)
```

```python
wine_pred = dt_wine.predict(w_X_train)
print(dt_wine.score(w_X_train, w_y_train))
print(classification_report(w_y_train, wine_pred))
```

```python
wine_pred = dt_wine.predict(w_X_test)
print(dt_wine.score(w_X_test, w_y_test))
print(classification_report(w_y_test, wine_pred))
```

```python
cancer_pred = dt_cancer.predict(c_X_train)
print(dt_cancer.score(c_X_train, c_y_train))
print(classification_report(c_y_train, cancer_pred))
```

```python
cancer_pred = dt_cancer.predict(c_X_test)
print(dt_cancer.score(c_X_test, c_y_test))
print(classification_report(c_y_test, cancer_pred))
```
