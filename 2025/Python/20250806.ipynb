train.csv [파일] :
ID : 샘플별 고유 ID
age : 고객 나이
gender : 고객 성별
tenure : 고객이 서비스를 이용한 총 기간 (월)
frequent : 고객의 서비스 이용일
payment_interval : 고객의 결제 지연일
subscription_type : 고객의 서비스 등급
contract_length : 고객의 서비스 계약 기간
after_interaction : 고객이 최근 서비스 이용으로부터 경과한 기간 (일)
support_needs : 고객의 지원 필요도 (0 : 낮음 , 1 : 중간 , 2 : 높음)

ID : 샘플별 고유 ID

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('dacon/train.csv')
test = pd.read_csv('dacon/test.csv')
print(df.shape, test.shape)
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
# 빈데이터가 없음
```

```python
df.describe()
```

```python
# 안쓰는 컬럼 삭제
df.drop('ID', axis=1, inplace=True)
test.drop('ID', axis=1, inplace=True)
```

```python
# 여성은 0 남성은 1로 설정
df['gender'] = df.gender.apply(lambda x: 0 if x == 'F' else 1)
df.head()
```

```python
test['gender'] = test.gender.apply(lambda x: 0 if x == 'F' else 1)
test.head()
```

```python
df.subscription_type.unique(), test.subscription_type.unique() # 동일한 구독형 타입
```

```python
# 원핫인코딩
df_type = pd.get_dummies(df.subscription_type, dtype=int)
test_type = pd.get_dummies(test.subscription_type, dtype=int)
df_type.head()
```

```python
test_type.head()
```

```python
df2 = pd.concat([df, df_type], axis=1) # 원핫인코딩된 컬럼 병합
test2 = pd.concat([test, test_type], axis=1)
df2.head()
```

```python
df2.drop('subscription_type', axis=1, inplace=True)
test2.drop('subscription_type', axis=1, inplace=True)
```

```python
abs(df2.corr()['support_needs']).sort_values() # 거의 컬럼들간의 연관성이 없어보임
```

```python
data = df2.drop('support_needs', axis=1)
target = df2.support_needs
data.shape, target.shape
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(data, target, test_size=.3, random_state=42)
X_train.shape, X_test.shape
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
```

```python
dt.score(X_train, y_train)
```

```python
dt.score(X_test, y_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier()
knn.fit(X_train, y_train)
knn.score(X_train, y_train)
```

```python
knn.score(X_test, y_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)
rf.score(X_train, y_train)
```

```python
rf.score(X_test, y_test)
```

```python
from sklearn.svm import SVC
svm = SVC(random_state=42)
svm.fit(X_train, y_train)
svm.score(X_train, y_train)
```

```python
svm.score(X_test, y_test)
```

```python
submission = pd.read_csv('dacon/sample_submission.csv')
submission.head()
```

```python
test2
```

```python
res = rf.predict(test2)
```

```python
res[:5]
```

```python
from xgboost import XGBClassifier
xgb = XGBClassifier()
xgb.fit(X_train, y_train)
xgb.score(X_train, y_train)
```

```python
xgb.score(X_test, y_test)
```

```python
from lightgbm import LGBMClassifier
lgbm = LGBMClassifier(random_state=42)
lgbm.fit(X_train, y_train)
x_res = lgbm.predict(X_train)
sum(x_res == y_train)/len(y_train)
```

```python
res = lgbm.predict(test2)
res.shape, res[:5]
```

```python
res = xgb.predict(test2)
res[:5]
```

```python
submission['support_needs'] = res
submission
```

```python
# submission.to_csv('test_xg.csv', index=False)
submission.to_csv('test_lgbm.csv', index=False)
```

```python
# 테스트용 - 완료 후 삭제
from sklearn.metrics import f1_score

pub = pd.read_csv('public.csv')
pri = pd.read_csv('private.csv')

pub_score = submission[submission['ID'].isin(pub['ID'])]
pri_score = submission[submission['ID'].isin(pri['ID'])]

print(f1_score(pub['support_needs'], pub_score['support_needs'], average='macro'))
print(f1_score(pri['support_needs'], pri_score['support_needs'], average='macro'))
```
