```python
# BMI 2만명 데이터 만들기
def calc_bmi(h, w): # 키와 몸무게
    bmi = w / (h/100) ** 2
    return 'thin' if bmi < 18.5 else 'fat' if bmi > 25 else 'normal'
```

```python
# 데이터 파일
with open('bmi20000.csv', 'w', encoding='utf-8') as fp:
    fp.write('height,weight,label\n')
    cnt = {'thin': 0, 'normal': 1, 'fat': 2}
    for i in range(20000):
        h = np.random.randint(120, 200)
        w = np.random.randint(35, 120)
        label = calc_bmi(h,w)
        cnt[label] += 1
        fp.write('{0},{1},{2}\n'.format(h,w,label))
print("OK")
```

```python
import pandas as pd
# min-max scaling
def minmax(data:pd.DataFrame):
    data = (data - data.min()) / data.max() - data.min()
    return data
```

```python
print('변환 전: \n',titanic['Fare'].head())
```

```python
minmax(titanic.Fare)
```

```python
titanic.Fare.min(), titanic.Fare.max()
```

```python
def stdscale(data):
    data = (data-data.mean()) / data.std()
    return data
stdscale(titanic.Fare)
```

```python
fare = minmax(titanic.Fare)
fare
```

```python
fare.min(), fare.max()
```

```python
# 표준화: 분산1, 평균 0으로 만들어주는 작업
# minmax: 최소 0 최대 1로 만들어주는 작업
from sklearn.preprocessing import MinMaxScaler, StandardScaler
mm = MinMaxScaler() # min-max
std = StandardScaler() # standart(정규화)
mm_sc = mm.fit_transform(titanic[['Fare']])
std_sc = std.fit_transform(titanic[['Fare']])
df = pd.DataFrame(mm_sc, titanic.Fare)
df['standard_fare'] = std_sc
df2 = df.reset_index()
df2.columns = ['origin_fare','minmax_fare', 'standard_fare'] 
df2
```
