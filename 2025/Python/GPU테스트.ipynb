```python
import tensorflow as tf
import torch
import keras
import platform
import numpy as np
from keras import layers, models, config as kcfg

print("=== Python 환경 ===")
print("Python:", platform.python_version())

print("\n=== TensorFlow ===")
print("버전:", tf.__version__)
gpus = tf.config.list_physical_devices('GPU')
print("GPU 장치 목록:", gpus)

if gpus:
    with tf.device("/GPU:0"):
        a = tf.random.normal([1000, 1000])
        b = tf.matmul(a, a)
        print("TensorFlow 연산 결과 shape:", b.shape)
else:
    print("TensorFlow에서 GPU를 찾을 수 없음")

print("\n=== PyTorch ===")
print("버전:", torch.__version__)
print("CUDA 사용 가능:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU 이름:", torch.cuda.get_device_name(0))
    x = torch.randn(2000, 2000, device="cuda")
    y = x @ x
    print("PyTorch 연산 결과 shape:", y.shape, "디바이스:", y.device)
else:
    print("PyTorch에서 GPU를 찾을 수 없음")

print("\n=== Keras (TensorFlow backend) ===")
print("버전:", keras.__version__)
print("백엔드:", keras.backend.backend())

model = models.Sequential([
    layers.Input(shape=(1000,)),  # 경고 방지: Input 레이어 사용
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])
model.compile(optimizer="adam", loss="categorical_crossentropy")

x_data = np.random.rand(500, 1000).astype("float32")
y_data = np.random.rand(500, 10).astype("float32")
model.fit(x_data, y_data, epochs=1, batch_size=32, verbose=0)
print("Keras(TensorFlow) 학습 1회 완료")
```
