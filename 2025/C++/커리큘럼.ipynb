## 🔍 1. "현대식 C++"인지 알 수 있는 기준

### ✅ 현대 C++ 코드의 특징

| 항목                            | 예시 또는 특징                                         |
| ----------------------------- | ------------------------------------------------ |
| **STL 컨테이너 사용**               | `std::vector`, `std::string`, `std::map` 등       |
| **스마트 포인터 사용**                | `std::unique_ptr`, `std::shared_ptr`             |
| **자동 타입 추론**                  | `auto` 키워드                                       |
| **범위 기반 for문**                | `for (auto& val : vec)`                          |
| **람다 함수 사용**                  | `[&](int x){ return x*x; }`                      |
| **헤더만 사용**                    | `<vector>`, `<memory>`, `<iostream>` 등           |
| **`new`/`delete` 대신 스마트 포인터** | `std::make_unique<T>()`, `std::make_shared<T>()` |

→ 위 요소들을 쓰고 있다면 \*\*현대 C++\*\*이라 할 수 있어요. (특히 C++11, 14, 17 이후 등장 기능들)

## 📌 공부 방법 요약 (추천 루트)

1. **벡터/배열 + 포인터 개념을 확실히 잡기** → 연습 문제 위주
